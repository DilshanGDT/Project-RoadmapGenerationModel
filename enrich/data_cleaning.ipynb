```python
import json
import re
import pandas as pd
from tqdm import tqdm
import numpy as np

input_file = "sri_lanka_safaris.json"   # your scraped data

with open(input_file, "r", encoding="utf-8") as f:
    raw_data = json.load(f)

print("Total records loaded:", len(raw_data))
```

```python
def clean_price(price):
    if not price or price == "":
        return None
    price = price.replace(",", "").strip()
    
    # Extract first number
    match = re.search(r"\d+", price)
    if match:
        return int(match.group())
    return None
```

```python
def clean_rating(r):
    if not r: 
        return None
    match = re.search(r"\d+(\.\d+)?", r)
    if match:
        return float(match.group())
    return None
```

```python
def clean_reviews(r):
    if not r:
        return 0
    r = r.replace(",", "")
    match = re.search(r"\d+", r)
    if match:
        return int(match.group())
    return 0
```

```python
def clean_duration(text):
    if not text:
        return None
    
    text = text.replace("–", "-")  # normalize dash
    text = text.strip()
    return text
```

```python
def clean_text(t):
    if not t:
        return ""
    t = re.sub(r"<.*?>", " ", t)  # remove HTML tags
    t = re.sub(r"\s+", " ", t)
    return t.strip()
```

```python
def clean_safari_item(item):
    return {
        "title": clean_text(item.get("title", "")),
        "image_links": item.get("image_links", []),
        "rating": clean_rating(item.get("rating", "")),
        "total_reviews": clean_reviews(item.get("total_reviews", "")),
        "recommendation": clean_text(item.get("recommendation", "")),
        "type": clean_text(item.get("type", "")),
        "travel_duration": clean_duration(item.get("travel_duration", "")),
        "description": clean_text(item.get("description", "")),
        "starting_price": clean_price(item.get("starting_price", "")),
    }
```

```python
cleaned_data = []

for item in tqdm(raw_data, desc="Cleaning data"):
    cleaned = clean_safari_item(item)
    cleaned_data.append(cleaned)

print("Cleaning complete!")
```

```python
output_file = "safari_cleaned.json"

with open(output_file, "w", encoding="utf-8") as f:
    json.dump(cleaned_data, f, indent=2, ensure_ascii=False)

print("Saved cleaned data to:", output_file)
```
